```python
import pandas as pd
df = pd.read_csv('ergebnisse_ukraine_fdp_kategorien_breit_pro_segment.csv', sep=';')
df
```

```python
df_absolut_ja = df[df['stance'] == 'ABSOLUT JA%']
df_absolut_ja
```

```python
# Die maximale Spaltenbreite auf eine feste Zeichenanzahl (z.B. 100) erhöhen
#pd.set_option('display.max_colwidth', 100)
df_absolut_ja.style.set_properties(subset=['explanation'], **{'width': '100px'})

df_absolut_ja = df_absolut_ja.iloc[:3]

# DataFrame anzeigen
df_absolut_ja
```

```python
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages
import textwrap

# 1. Funktion für den automatischen Zeilenumbruch definieren
def wrap_text(text, width=50):
    return '\n'.join(textwrap.wrap(str(text), width=width))

# 2. Kopie des DataFrames erstellen und Textspalten sauber umbrechen
df_wrapped = df_absolut_ja.copy()
#df_wrapped['document'] = df_wrapped['document'].apply(lambda x: wrap_text(x, width=40))
df_wrapped['explanation'] = df_wrapped['explanation'].apply(lambda x: wrap_text(x, width=55))

# 3. PDF-Export als Vektorgrafik starten
with PdfPages('dataframe_vektor_tabelle.pdf') as pdf:
    # Breitformat-Figur erstellen (wichtig für die Spaltenaufteilung)
    fig, ax = plt.subplots(figsize=(10, 4))
    ax.axis('tight')
    ax.axis('off')
    
    # Exakt 4 Breiten für die 4 echten Spalten (Index wird autom. berechnet)
    column_widths = [0.35, 0.12, 0.08, 0.45]
    
    # Tabelle generieren
    table = ax.table(cellText=df_wrapped.values, 
                     colLabels=df_wrapped.columns, 
                     rowLabels=df_wrapped.index,
                     loc='center', 
                     cellLoc='left',
                     colWidths=column_widths)
    
    # Schriftgröße fixieren und Zeilenhöhe (scale) für den umgebrochenen Text anpassen
    table.auto_set_font_size(False)
    table.set_fontsize(8)
    table.scale(1.0, 4.8)
    
    # Als Vektor-PDF in die Datei schreiben
    pdf.savefig(fig, bbox_inches='tight')
    plt.close()
```
