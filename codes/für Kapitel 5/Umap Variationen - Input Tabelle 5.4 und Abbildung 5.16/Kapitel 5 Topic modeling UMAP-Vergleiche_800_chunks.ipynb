```python
import pandas as pd

anzahl_cluster = pd.Series(["56", "312", "230", "204"])
anzahl_outlier = pd.Series(["13674", "15972", "13479", "13122"])
anzahl_outlier_reduced = pd.Series(["9978", "3082", "3142", "3248"])
UMAP_neighbor = pd.Series(["8", "15", "30", "40"])
UMAP_components = pd.Series(["5", "4", "2", "2"])
hdbscan_cluster = pd.Series(["12", "12", "15", "15"])
hdbscan_min_samples = pd.Series(["4", "4", "4", "4"])
    
# DataFrame direkt aus den erstellten Series zusammensetzen
df = pd.DataFrame({
    'Cluster': anzahl_cluster,
    'Outlier': anzahl_outlier,
    'Outlier reduced': anzahl_outlier_reduced,
    'UMAP neighbor': UMAP_neighbor,
    'UMAP components': UMAP_components,
    'Hdbscan Clustergröße': hdbscan_cluster,
    'Hdbscan min samples': hdbscan_min_samples
})

df = df.reset_index(drop=True)
df
```

```python
# in LaTex überführen

latex_code = df.to_latex(index=False, caption='umap_vergleiche', label='tab:umap_vergleiche',column_format='cc|c|c|c|c')
with open('tabelle_umap_vergleiche.tex', 'w') as f:
    f.write(latex_code)
```
